```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('advertising.csv')
df
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
```

```python
np.random.seed(0)
```

```python
df_train, df_test = train_test_split(df, train_size=0.7, test_size=0.3, random_state=100)
```

<h1 style="color: brown;">Selecting the list of variables for scaling</h1>

```python
num_vars = ['TV', 'Radio', 'Newspaper']
```

<h2 style="color: brown;">Training and Testing</h2>

```python
y_train = df_train.pop('Sales')
y_test = df_test.pop('Sales')
```

<h2 style="color: brown;">Scaling</h2>

```python
scaler = MinMaxScaler()
```

```python
df_train[num_vars] = scaler.fit_transform(df_train[num_vars])
```

```python
df_test[num_vars] = scaler.transform(df_test[num_vars])
```

```python
import statsmodels.api as sm
```

```python
X_train_sm = sm.add_constant(df_train)
X_test_sm = sm.add_constant(df_test)
```

```python
model = sm.OLS(y_train, X_train_sm).fit()
```

```python
y_pred = model.predict(X_test_sm)
```

```python
print(model.summary())
```

### Scaling the Target variable 'Sales' and calculating the accuracy

```python
target_scaler = MinMaxScaler()
y_train_scaled = target_scaler.fit_transform(y_train.values.reshape(-1, 1))
y_test_scaled = target_scaler.transform(y_test.values.reshape(-1, 1))
```

```python
model_sc = sm.OLS(y_train_scaled, X_train_sm).fit()
```

```python
print(model_sc.summary())
```

### Calculating the testing accuracy

```python
model_test=sm.OLS(y_test, X_test_sm).fit()
```

```python
print(model_test.summary())
```

```python
model_test_scaled=sm.OLS(y_test_scaled, X_test_sm).fit()
```

```python
print(model_test_scaled.summary())
```
